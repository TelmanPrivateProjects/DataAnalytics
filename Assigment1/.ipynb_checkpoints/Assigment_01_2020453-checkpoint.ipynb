```python
## task3.2

# Assignings lists
list1 = [1, 2, 3]
list2 = [3, 4, 5]

# Converting lists to tuples
tuple1 = tuple(list1)
tuple2 = tuple(list2)

# Using set intersection method to find common elements
common_elements = list(set(tuple1).intersection(tuple2))

# Print result
if common_elements:
    print("Common elements:", common_elements)
else:
    print("No common elements found.")
```

### 3.4 Explanation of Keys and Values:
Keys in the parent dictionary (top_books):"book1", "book2", "book3", "book4", "book5".
Values in the parent dictionary: Each value is child dictionary representing details of the book.

Keys in the child dictionaries:"title", "author", "published".
Values in the child dictionaries: The actual information about the book (title and author and published year).

### 3.4 Why this structure?:

Dictionaries are ordered and changeable collections that store data using unique keys. This makes them ideal for organizing books with clear keys like "title" and "author." Unlike lists or tuples, dictionaries label data with keys instead of just positions, and unlike sets, they keep order and allow key-value pairs without duplicates.



```python
## task3.4

# Assigning child dictionaries
book1 = {
    "title": "To Kill a Mockingbird",
    "author": "Harper Lee",
    "published": 1960
}

book2 = {
    "title": "The Lord of the Rings",
    "author": "J.R.R. Tolkien",
    "published": 1955
}

book3 = {
    "title": "The Little Prince",
    "author": "Antoine de Saint-Exupéry",
    "published": 1943
}

book4 = {
    "title": "1984",
    "author": "George Orwell",
    "published": 1949
}

book5 = {
    "title": "Pride and Prejudice",
    "author": "Jane Austen",
    "published": 1813
}

# Assigning parent dictionary for childs
top_books = {
    "book1": book1,
    "book2": book2,
    "book3": book3,
    "book4": book4,
    "book5": book5
}
```

```python
## task3.5

for book in top_books.values():
    book["language"] = "English"

# Printing parent dictionary
print(top_books)
```

### 5.4

#### At least 3 issues regarding dataset:

There are duplicate rows that need to be removed.
Some rows have zero values in the value column, which should be cleaned out. 
The quantity column contains non-numeric text values that must be fixed or removed.
There are missing values (NaNs) that need to be dropped for clean data. 
After removing row, indexing should be fix.
Column types are error-prone.

#### Cleaning dataset
Firstly removed duplicate rows to keep only unique entries.
Secondly, filtered out rows where the value column is zero.
Converted the quantity column to numbers, turning invalid text into NaN, then removes those rows.
Dropped all rows that contain any missing (NaN) values.

Finally, reseted the index and changes columns to their appropriate data types for better efficiency and accuracy.

```python
## task5.4

# Importing pandas library
import pandas as pd

# Importing the CSV file
df = pd.read_csv("inventory.csv")

# Cheking null values
print(df.info())

# Cleaning duplicates
df = df.drop_duplicates()

# Cleaning 0 values in value column
df = df[df['value'] != 0]

# Cleaning non numeric values in quantity column
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')
df = df[df['quantity'] != "AA"]
df = df[df['quantity'] != "BB"]
df = df[df['quantity'] != "CC"]

# Cleaning NaN values from df
df = df.dropna() 

# Cleaning index after deletions
df = df.reset_index(drop=True)

# Converting to appropriate types
df = df.astype({
    'product.partNumber': 'string',
    'location.locationIdentifier': 'string',
    'inventoryType': 'category',
    'quantity': 'int',
    'quantityUnits': 'category',
    'value': 'float',
    'valueCurrency': 'category',
    'plannerCode': 'category',
    'velocityCode': 'category',
    'inventoryParentType': 'category',
    'class': 'category',
    'segment': 'category'
})

# Printing result
print(df.info())
print(df.to_string())

```
