### Task 1 – Normalize the Orders Table

The original `orders` table stores one book per order, which is inefficient and not scalable. To resolve this issue, we normalized the data into two separate tables:

1. **orders_table** – Contains unique information for each order such as `order_id`, `order_date`, `customer_id`, and `payment_method`.

2. **order_items_table** – Links each order to the books it contains, including `book_id` and `quantity`.

```python
# task1

# Importing pandas library
import pandas as pd

# Reading raw data
orders = pd.read_csv("data/orders.csv")

# Creating a normalized orders_table
orders_table = orders[['order_id', 'order_date', 'customer_id', 'payment_method']].drop_duplicates()

# Creating a normalized 'order_items' table
order_items_table = orders[['order_id', 'book_id', 'quantity']].copy()

# Adding a primary key to the order_items table for better reference
order_items_table['order_item_id'] = range(1, len(order_items_table) + 1)

# Reordering columns for clarity
order_items_table = order_items_table[['order_item_id', 'order_id', 'book_id', 'quantity']]

# Displaying the normalized tables
print("\nOrders Table:")
display(orders_table)

print("\nOrder Items Table:")
display(order_items_table)
```

### Task 3 – Increase Prices for Fiction Category

Joined the `books` and `categories` tables using the `category_id` foreign key. After identifying all books in the **Fiction** category, increased their prices by **15%**.


```python
# task3

# Importing pandas library
import pandas as pd

# Loading raw data
books = pd.read_csv("data/books.csv")
categories = pd.read_csv("data/categories.csv")

# Merging books with categories on 'category_id'
books_with_category = pd.merge(books, categories, on='category_id', how='left')

# Increasing the price by 15% only for 'Fiction' category
books_with_category.loc[books_with_category['category_name'] == 'Fiction', 'price'] *= 1.15

# Display updated fiction books
fiction_books = books_with_category[books_with_category['category_name'] == 'Fiction']
display(fiction_books[['title', 'price']])
```

### Task 6 – Average Rating for Orders on 19.11.23

Filtered all orders that were placed on **November 19, 2023**, and extracted the associated `book_id`s. 
Then, looked up the reviews related to these books and calculated the average rating.

```python
# task 6

# Importing pandas library
import pandas as pd

# Loading raw data
orders = pd.read_csv("data/orders.csv")
reviews = pd.read_csv("data/reviews.csv")

# Ensuring date format is consistent
orders['order_date'] = pd.to_datetime(orders['order_date'], format='%d.%m.%y')

# Filtering orders placed on 19.11.2023
target_date = pd.to_datetime('19.11.23', format='%d.%m.%y')
orders_on_target_date = orders[orders['order_date'] == target_date]

# Getting all unique book_ids from orders on 19.11.2023
book_ids = orders_on_target_date['book_id'].unique()

# Filtering reviews for those books
reviews_for_books = reviews[reviews['book_id'].isin(book_ids)]

# Calculating average rating
average_rating = reviews_for_books['rating'].mean()

print(f"Average review rating for books ordered on 19.11.23: {average_rating:.2f}")
```

### Task 8 – Providing Date, Customer Name, Email, and Ordered Books of the Most Recent Orders

In this task, the most recent order date was determined from the `orders` table.  
Then, all orders placed on that date were filtered. These were joined with the `customers` and `books` tables  
to obtain customer names, email addresses, and book titles.

```python
# task8

# Importing pandas library
import pandas as pd

# Load raw data
orders = pd.read_csv("data/orders.csv")
customers = pd.read_csv("data/customers.csv")
books = pd.read_csv("data/books.csv")

# Converting 'order_date' to datetime
orders['order_date'] = pd.to_datetime(orders['order_date'], format='%d.%m.%y')

# Finding the most recent order date
latest_date = orders['order_date'].max()

# Filtering orders made on the most recent date
latest_orders = orders[orders['order_date'] == latest_date]

# Merging orders with customers to get customer name and email
latest_orders_with_customers = pd.merge(latest_orders, customers, on='customer_id', how='left')

# Merging with books to get book title
latest_orders_full = pd.merge(latest_orders_with_customers, books, on='book_id', how='left')

# Selecting relevant columns to display
result = latest_orders_full[['order_date', 'name', 'email', 'title']]

# Displaying the result
display(result)
```

### Task 9 – Identifying the Most Popular Book Based on Order Frequency

In this task, the number of times each book was ordered was calculated by grouping the `orders` table by `book_id`.  
The resulting counts were sorted in descending order, and the top result was selected.

To make the output more interpretable, the most ordered book ID was joined with the `books` table  
to retrieve its title and additional details.

> **Note:** Book IDs 48 and 79 are tied with 7 orders each. A stable sort is used so the result is deterministic (the lower `book_id` is returned).

```python
# task9

# Importing pandas library 
import pandas as pd

# Reading datasets
orders = pd.read_csv("data/orders.csv")
books = pd.read_csv("data/books.csv")

# Grouping by book_id and counting the number of orders
book_order_counts = orders.groupby('book_id').size().reset_index(name='order_count')

# Sorting in descending order to find the most ordered book
most_ordered = book_order_counts.sort_values(by='order_count', ascending=False, kind='stable').head(1)

# Merging with books to get the book title and details
most_popular_book = pd.merge(most_ordered, books, on='book_id', how='left')

# Displaying the result
display(most_popular_book[['book_id', 'title', 'order_count']])
```

### Task 10 – Identifying Customers Who Spent More Than 100 Euros

In this task, the total amount spent by each customer was calculated by grouping the `orders` table by `customer_id`  
and summing the `total_price` values.

Customers with total spending above 100 euros were filtered.  
To provide meaningful output, their names and email addresses were added by joining the `customers` table.

```python
# task10

# Importing pandas library
import pandas as pd

# Reading datasets
orders = pd.read_csv("data/orders.csv")
customers = pd.read_csv("data/customers.csv")

# Grouping by customer_id and calculating total spending
customer_spending = orders.groupby('customer_id')['total_price'].sum().reset_index()

# Filtering customers who spent more than 100 euros
high_spenders = customer_spending[customer_spending['total_price'] > 100]

# Merging with customers to get name and email
high_spenders_details = pd.merge(high_spenders, customers, on='customer_id', how='left')

# Displaying relevant columns
display(high_spenders_details[['customer_id', 'name', 'email', 'total_price']])
```

### Task 12 – Splitting the Customer Name into First and Last Name

In this task, the `name` column from the `customers` table was split into two separate columns: `last_name` and `first_name`.  
It was assumed that the original name format is `"LastName, FirstName"` (with a comma separating the two parts).

The `.str.split()` function in pandas was applied with the delimiter set as comma.  
The result was expanded into two new columns using the `expand=True` argument.  
Whitespace was removed using `.str.strip()` to clean the resulting fields.


```python
# task12

# Importing pandas library
import pandas as pd

# Reading customer data
customers = pd.read_csv("data/customers.csv")

# Splitting the name by the first space only (first_name, last_name)
name_split = customers['name'].str.split(' ', n=1, expand=True)

# Renaming the new columns
customers['first_name'] = name_split[1]
customers['last_name'] = name_split[0]

# Displaying result
display(customers[['customer_id', 'name', 'first_name', 'last_name']])
```

### Task 15 – Restructuring the Book-Category Relationship (Many-to-Many)

Originally, each book in the system was associated with only one category using a `category_id` field within the `books` table.  
While this simplified the database schema, it failed to reflect real-world complexity where books can belong to multiple genres.

To resolve this limitation, the data model was restructured by introducing a new intermediate table: `book_categories`.  
This new table allows for a many-to-many relationship between books and categories by storing pairs of `book_id` and `category_id`.

In order to test the revised structure, three books written by 'J.K. Rowling' were added and linked to multiple categories.  
Specifically, each of her books was associated with both the 'Fiction' and 'Mystery' categories.

```python
# task15

# Importing pandas library
import pandas as pd

# Load raw data
categories = pd.read_csv("data/categories.csv")
books = pd.read_csv("data/books.csv")

# Creating J.K. Rowling test books
rowling_books = pd.DataFrame({
    'book_id': [901, 902, 903],
    'title': ['The Ickabog', 'Troubled Blood', 'The Casual Vacancy'],
    'author': ['J.K. Rowling'] * 3,
    'price': [14.99, 18.50, 16.00]
})

# Appending test books to the books DataFrame
books = pd.concat([books, rowling_books], ignore_index=True)

# Creating a new many-to-many book_categories table:
# Each Rowling book will be linked to Fiction (1) and Mystery (5)
book_categories = pd.DataFrame({
    'book_id': [901, 901, 902, 902, 903, 903],
    'category_id': [1, 5, 1, 5, 1, 5]
})

# Filtering books written by 'J.K. Rowling'
rowling_books = books[books['author'] == 'J.K. Rowling']

# Dropping category_id from dateset cause of future conflicts
rowling_books = rowling_books.drop(columns=['category_id'], errors='ignore')

# Merging Rowling books with book_categories
rowling_books_with_categories = pd.merge(rowling_books, book_categories, on='book_id', how='left')

# Merging with categories to get category names
rowling_books_full = pd.merge(rowling_books_with_categories, categories, on='category_id', how='left')

# Displaying only Rowling books with their categories
display(rowling_books_full[['book_id', 'title', 'category_name']])

```
